```python
import pandas as pd
from pathlib import Path

# 兼容 Jupyter / nbconvert / VSCode 多种执行环境
_notebook_dir = Path(__file__).parent if '__file__' in globals() else Path.cwd()
_search_bases = [
    _notebook_dir.parent,           # notebooks/
    Path.cwd(),
    Path.cwd().parent,
    Path('/d/codeproject/binwh.quant-pilot'),
    Path('/d/codeproject'),
]

FORECAST_REL = Path('data') / 'output' / 'sina_300910_forecast_summary.csv'
FORECAST_PATH = None
for base in _search_bases:
    candidate = base / FORECAST_REL
    if candidate.exists():
        FORECAST_PATH = candidate
        break

# 最终 fallback: 从所有 base 递归搜索
if FORECAST_PATH is None:
    for base in _search_bases:
        hits = list(base.rglob('sina_300910_forecast_summary.csv'))
        if hits:
            FORECAST_PATH = hits[0]
            break

print(f"数据文件: {FORECAST_PATH}")
forecast_df = pd.read_csv(FORECAST_PATH, encoding='utf-8-sig', index_col=0)
forecast_df
```
